# Rossby Waves

# Recap 

Rossby waves are excited when there are perturbations in the height and velocity fields around a **geostrophically** balanced initial state ($\text{Ro}\ll 1$), in a fluid where there is **differential rotation**, i.e. the Coriolis force experienced by a fluid parcel is spatially varying. In idealised geophysical fluid dynamic applications, this is usually studied by allowing the Coriolis parameter, $f$, to vary linearly with the $y$ (latitude-like) coordinate, i.e. $f = f_0 + \beta y$, where $\beta$ is constant (this is known as the *beta plane* approximation). In non-dimensionalised coordinates, this becomes $f = 1/\text{Ro} + \beta y$, where $\beta$ and $y$ are now dimensionless quantities.

### Geostrophic balance
The background state from which the waves are excited is geostrophic balance. This is governed by the equations:

$$
\begin{align}
-\text{Ro}^{-1}v &= -\partial_x h\\
+\text{Ro}^{-1}u &= -\partial_y h
\end{align}
$$

In a perfectly balanced initial state, the pressure gradient always acts at **right angles** to the velocity (which can be verified by taking $\mathbf{u}\cdot\nabla h$) and using the equations above.

<img src="./images/geostrophic_balance.png" alt="drawing" width="400"/>

### Rossby wave dispersion relation
It can be shown (see below for the full derivation) that perturbations around this initial state lead to disturbances with the dispersion relation:

$$
\omega = -\frac{\beta k_x}{k_x^2 + k_y^2 + \text{Ro}^{-2}}
$$
In the long wavelength limit, where $k^2 = \lambda^{-2} \ll \text{Ro}^{-2}$, the waves are non-dispersive waves that travel in the negative $x$ direction (i.e., westwards):
$$
\begin{align}
\omega = -\beta\text{Ro}^2 k_x.
\end{align}
$$
Note that since $\text{Ro} \ll 1$ for geostrophic balance, the wave speed is typically $\ll 1$, i.e. much smaller than the typical gravity wave speed.

### Physical interpretation
Our shallow water flow conserves the total voriticity:
$$
\frac{dq}{dt} = \frac{d}{dt}\left(\frac{(\nabla \times \mathbf{u})\cdot\hat{\mathbf{z}} + f}{h}\right) = 0
$$
The curl of the velocity, the "relative voriticity", is a measure of the fluid's local rotation. 

<img src="./images/vorticity.png" alt="drawing" width="200"/>

A parcel of fluid displaced in the positive $y$ direction will increase its value of $f$, since $f = \text{Ro}^{-1} + \beta y$. To conserve $q$, the parcel must reduce its relative vorticity (and spin clockwise). The opposite effect occurs if the parcel is displaced in the negative $y$ direction. The overall effect is that a waveform will be transported in the negative $x$ direction (see diagram below).

<img src="./images/rossby.png" alt="drawing" width="70%"/>


### Derivation (for interest, not examinable)
When $\text{Ro}\ll1$, the leading order flow is a divergence free, geostrophic flow. 
We will define the relative vorticity, $\zeta \equiv \partial_x v - \partial_y u$. It will also be useful to define a streamfunction, $\psi$, such that the leading order geostrophic flow is defined by:
$$
\begin{align}
u &\approx -\partial_y \psi \\
v &\approx +\partial_x \psi
\end{align}
$$

Note by defining the streamfunction as such, we get $\zeta = \nabla^2 \psi$. From this definition and the definition of geostrophic balance, we can also write:
$$
\eta = \frac{\psi}{\text{Ro}}
$$
where $\eta$ is the height perturbation, $\eta = h - 1$.
The linearised shallow water equations are:
$$
\begin{align}
\partial_t \eta + \partial_x u + \partial_y v &= 0 \\
\partial_t u - (\text{Ro}^{-1}+\beta y) v + \partial_x \eta &= 0 \\
\partial_t v + (\text{Ro}^{-1}+\beta y) u + \partial_y \eta & = 0
\end{align}
$$
We start by constructing $\partial_t \zeta$ by taking the $x$ derivative of the third equation above and subtracting the $y$ derivative of the second equation from it. We find:

$$
\partial_t \zeta + (\text{Ro}^{-1}+\beta y)(\partial_x u + \partial_y v) + \beta v = 0
$$

From the first equation, $\partial_x u + \partial_y v = \partial_t \eta = -\partial_t\psi/\text{Ro}$. The term $\beta y \nabla \cdot u$ can also be neglected, because $\text{Ro}^{-1} \gg \beta y$ and the divergence term is also very small (note the leading order geostrophic flow is divergence free). If we also substitute $\zeta = \nabla^2 \psi$, we get:
$$
\partial_t\left(\nabla^2\psi - \frac{\psi}{\text{Ro}^2}\right) + \beta \partial_x \psi = 0,
$$
Substituting in a plane wave of form $\psi = \psi_0 \exp\left(i(\omega t - k_x x - k_y y)\right)$, we recover the dispersion relation:
$$
\begin{align}
\omega &= -\frac{\beta k_x}{k_x^2 + k_y^2 + \text{Ro}^{-2}}\\
\end{align}
$$


### Alternative derivation from conservation of potential vorticity

Start with a fluid with potential vorticity:

$$
q = \frac{\zeta + \text{Ro}^{-1} + \beta y}{h} = \frac{\zeta + \text{Ro}^{-1} + \beta y}{\eta + 1}\approx (\zeta + \text{Ro}^{-1} + \beta y)(1 - \eta)
$$

Expanding and neglecting the term in $\eta \beta y$, which is small:

$$
q \approx \zeta + \text{Ro}^{-1} - \text{Ro}^{-1}\eta + \beta y
$$

This is conserved in flow, i.e.

$$
\partial_t q + u \partial_x q + v\partial_y q = 0
$$

Substituting our $q$ in, we get:

$$
\partial_t\left(\zeta - \frac{\eta}{\text{Ro}}\right) + \beta v = 0,
$$
where all terms quadratic in $u$ or $v$ have been neglected (i.e., inertial terms are small relative to the pressure gradient force). Then using our streamfunction definitions, $\eta = \psi/\text{Ro}$, $\zeta = \nabla^2 \psi$, $v = \partial_x \psi$, we recover the same equation as above.
